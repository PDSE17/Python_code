TYPES OF CLUSTERING

1.K-MEANS
---> most popular clustering algorithm
---> it divides into k clusters
k ---> no. of clusters

k=3 3 groups mei divide karke 


```python
# clustering 
import pandas as pd
from sklearn.cluster import KMeans

data = {
    "age":[22,25,24,45,50,48,20,47],
    "income":[22000,25000,24000,45000,50000,48000,20000,47000]
}

df = pd.DataFrame(data)
print(df)

# input | feature
x = df[["age","income"]]

# model
model = KMeans(n_clusters=2,random_state=42)

model.fit(x)

# clusters
clusters = model.labels_
print(clusters)
df["cluster"] = clusters
print(df)
```

```python
# clustering another example
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

data = {
    "age":[22,25,24,45,50,48,20,47],
    "income":[22000,25000,24000,45000,50000,48000,20000,47000] 
}

df = pd.DataFrame(data)
print(df)

x = df[["age","income"]]

model = KMeans(n_clusters=2,random_state=42)
model.fit(x)

df["cluster"] = model.labels_

plt.scatter(df["age"],df["income"],c=df["cluster"])
plt.xlabel("age")
plt.ylabel("income")
plt.title("group the data")
plt.show()
```

```python
# clustering another example
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

data = {
    "age":[22,13,21,42,47,65,12,34],
    "voting rights":[1,0,1,1,1,1,0,1]
}

df = pd.DataFrame(data)
print(df)

x = df[["age","voting rights"]]

model = KMeans(n_clusters=2,random_state=42)
model.fit(x)

df["cluster"] = model.labels_

plt.scatter(df["age"],df["voting rights"],c=df["cluster"])
plt.xlabel("age")
plt.ylabel("voting rights")
plt.title("group the data")
plt.show()
```
